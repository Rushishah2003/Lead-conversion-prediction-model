# Exploratory Data Analysis (Lead Conversion)

**Objective**: Understand the dataset patterns to inform feature engineering and model selection.
**Author**: AI/ML Engineer

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Data
df = pd.read_csv('../leads_dataset.csv')
print(f"Dataset Shape: {df.shape}")
df.head()
```

## 1. Target Imbalance Analysis

```python
conversion_rate = df['converted'].mean()
print(f"Conversion Rate: {conversion_rate:.1%}")

sns.countplot(x='converted', data=df)
plt.title('Target Variable Distribution (Converted vs Not)')
plt.show()
```

> **Observation**: 29% Conversion. Imbalanced, but not extreme (like Fraud at 0.1%).
> **Decision**: We should use `StratifiedShuffleSplit` for validation to maintain this ratio.

## 2. Missing Values check

```python
missing = df.isnull().sum()
missing[missing > 0]
```

```python
# Check Response Time Distribution to decide on Mean vs Median
sns.histplot(df['response_time_hours'].dropna())
plt.title('Response Time Distribution')
plt.show()

print(f"Mean: {df['response_time_hours'].mean():.2f}")
print(f"Median: {df['response_time_hours'].median():.2f}")
```

> **Observation**: `response_time_hours` is right-skewed. The Mean is pulled higher by slow responses.
> **Decision**: Impute missing values with **Median** to be robust.

## 3. Investigating Company Size Impact

```python
size_order = ['1-10', '11-50', '51-200', '201-500', '501-1000', '1000+']
plt.figure(figsize=(10, 6))
sns.barplot(x='company_size', y='converted', data=df, order=size_order)
plt.title('Conversion Rate by Company Size')
plt.show()
```

> **Insight**: Larger companies seem to convert at a significantly higher rate.
> **Decision**: Treat this feature as **Ordinal** (Tier 1 to 6) rather than just categorical.

## 4. Engagement Metrics Analysis

```python
metrics = ['website_visits', 'email_opens', 'form_submissions']
df.groupby('converted')[metrics].mean()
```

> **Observation**: Converted users have higher activity across the board. Form submissions are especially strong indicators.
> **Decision**: Create an aggregate `Engagement Score` feature.
